```python
#defining a function
def greet():
    print("Hello")

def greetName(name, surname="Unknown"):
    print(f"Hello {name} {surname}")

# calling a function
greet()
greetName(name = "Anuj")
```

From a function's perspective:
1) A parameter is the variable listed inside the parentheses in the function definition.
2) An argument is the actual value that is sent to the function when it is called.

```python
#Functions with multiple parameters

def greet(name, location):
    print(f"hello {name} & what is the weather in {location}?")

greet(location= 'Pune', name= 'Anuj')
greet('Pune', 'Anuj')
greet(location = 'Pune')
```
